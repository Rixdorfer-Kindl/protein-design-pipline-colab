```python
#@title 🔻 合并四批结果 + 漏斗前两级 { display-mode: "form" }
!pip install -q biopython
import os, glob, numpy as np, pandas as pd
from Bio.PDB import PDBParser
from Bio.Data.IUPACData import protein_letters_3to1

base = "/content/drive/MyDrive/RFdiffusion"
#@markdown ### 筛选阈值
min_contacts = 100 #@param {type:"integer"}
min_hs_touched = 1 #@param {type:"integer"}
max_ca_dev = 1.0 #@param {type:"number"}

pdbp = PDBParser(QUIET=True)
def chain_seq(pdb, cid):
    ch = pdbp.get_structure("s", pdb)[0][cid]
    return "".join(protein_letters_3to1.get(r.get_resname().capitalize(), "X") for r in ch if r.id[0] == " ")

# ---------- RFD1 ----------
r1 = []
for g in ["group4", "group5"]:
    bb = pd.read_csv(f"{base}/RFD1/{g}/backbone_interface_summary.csv")
    bb["backbone"] = bb["file"].str.replace(".pdb", "", regex=False)
    bb["batch"] = g
    for i, row in bb.iterrows():
        pdb = glob.glob(f"{base}/RFD1/{g}/**/{row["backbone"]}.pdb", recursive=True)
        bb.loc[i, "pdb"] = pdb[0] if pdb else None
        if pdb:
            bb.loc[i, "binder_len"] = len(chain_seq(pdb[0], "A"))
            bb.loc[i, "target_seq_bb"] = chain_seq(pdb[0], "B")
    r1.append(bb)
bb1 = pd.concat(r1, ignore_index=True)

mp = []
for f in glob.glob(f"{base}/RFD1/*/fast_*/*/mpnn_only_results.csv"):
    mp.append(pd.read_csv(f))
mp = pd.concat(mp, ignore_index=True)
parts = mp["sequence"].str.split("/", n=1, expand=True)
mp["binder_seq"] = parts[0]
mp["target_seq_mpnn"] = parts[1] if parts.shape[1] > 1 else None
rfd1 = bb1.merge(mp[["backbone", "design_id", "mpnn_score", "binder_seq", "target_seq_mpnn"]], on="backbone", how="left")
rfd1["target_unchanged"] = rfd1["target_seq_mpnn"] == rfd1["target_seq_bb"]
rfd1["model"] = "RFD1"
rfd1["id"] = rfd1["backbone"] + "_d" + rfd1["design_id"].astype("Int64").astype(str)

# ---------- RFD3 ----------
r3 = []
for run in ["run1", "run2"]:
    a = pd.read_csv(f"{base}/RFD3/{run}/rfd3_interface_summary.csv")
    m = pd.read_csv(f"{base}/RFD3/{run}/rfd3_metrics.csv").drop(columns=["seed"], errors="ignore")
    d = a.merge(m, on="file"); d["batch"] = run; r3.append(d)
rfd3 = pd.concat(r3, ignore_index=True)
rfd3["model"] = "RFD3"
rfd3["id"] = rfd3["batch"] + "_" + rfd3["file"].str.extract(r"(model_\d+)")[0]
rfd3["backbone"] = rfd3["id"]
rfd3["target_unchanged"] = True  # RFD3 的 target 是固定输入

# ---------- 统一列、打标记 ----------
keep = ["model", "batch", "backbone", "id", "binder_len", "binder_seq", "target_unchanged",
        "min_dist", "n_atom_contacts", "hs_min_dist", "hs_touched"]
extra3 = ["max_ca_deviation", "n_chainbreaks", "n_clashing.interresidue_clashes_w_backbone",
          "n_clashing.interresidue_clashes_w_sidechain", "loop_fraction"]
allc = pd.concat([rfd1[keep], rfd3[keep + extra3]], ignore_index=True)

allc["pass_geom"] = (allc["n_atom_contacts"] >= min_contacts) & (allc["hs_touched"] >= min_hs_touched)
allc["pass_quality"] = np.where(allc["model"] == "RFD3",
    (allc["n_chainbreaks"] == 0) & (allc["n_clashing.interresidue_clashes_w_backbone"] == 0)
    & (allc["max_ca_deviation"] < max_ca_dev), True)
allc["has_seq"] = allc["binder_seq"].notna()

# ---------- 漏斗计数 ----------
def count(df): return pd.Series({
    "骨架/设计总数": df["backbone"].nunique(),
    "序列总数": int(df["has_seq"].sum()),
    "① 几何通过(骨架)": df[df["pass_geom"]]["backbone"].nunique(),
    "② +结构质量(骨架)": df[df["pass_geom"] & df["pass_quality"]]["backbone"].nunique(),
    "② +有序列且target未改(序列)": int((df["pass_geom"] & df["pass_quality"] & df["has_seq"] & df["target_unchanged"]).sum()),
})
display(allc.groupby(["model", "batch"]).apply(count))
display(allc.groupby("model").apply(count))

out = f"{base}/analysis"; os.makedirs(out, exist_ok=True)
allc.to_csv(f"{out}/master_all.csv", index=False)
allc[allc["pass_geom"] & allc["pass_quality"] & allc["has_seq"] & allc["target_unchanged"]] \
    .to_csv(f"{out}/candidates_for_AF.csv", index=False)
print("💾 已保存 master_all.csv 与 candidates_for_AF.csv 到", out)
print("RFD1 target 被 MPNN 改动的序列数:", int((~rfd1["target_unchanged"]).sum()), "/", len(rfd1))
```

```python
import os
import pandas as pd

base = "/content/drive/MyDrive/RFdiffusion"
analysis_dir = f"{base}/analysis"

# 1. 读取 master_all.csv 和 现有的 candidates_for_AF.csv
master_path = f"{analysis_dir}/master_all.csv"
candidates_path = f"{analysis_dir}/candidates_for_AF.csv"

if os.path.exists(master_path) and os.path.exists(candidates_path):
    master_df = pd.read_csv(master_path)
    cand_df = pd.read_csv(candidates_path)

    # 2. 筛选出 backbone 为 group4_5 的数据
    group4_5_data = master_df[master_df["backbone"] == "group4_5"]

    if len(group4_5_data) > 0:
        # 3. 排除已经在 candidates 里存在的记录，避免重复添加
        existing_ids = set(cand_df["id"].dropna().astype(str))
        new_inserts = group4_5_data[~group4_5_data["id"].astype(str).isin(existing_ids)]

        if len(new_inserts) > 0:
            # 4. 合并并保存
            updated_cand_df = pd.concat([cand_df, new_inserts], ignore_index=True)
            updated_cand_df.to_csv(candidates_path, index=False)
            print(f"✅ 成功将 {len(new_inserts)} 条 group4_5 的记录追加到了 candidates_for_AF.csv 中！")
            print(f"目前 candidates 表中总行数: {len(updated_cand_df)}")
            display(updated_cand_df.tail(len(new_inserts) + 2))
        else:
            print("⚠️ group4_5 的数据已经存在于 candidates 表格中，未进行重复添加。")
    else:
        print("❌ 未在 master_all.csv 中检索到 backbone 为 'group4_5' 的数据。")
else:
    print("❌ 找不到 master_all.csv 或 candidates_for_AF.csv，请先确保运行了上方的合并筛选单元格！")
```
